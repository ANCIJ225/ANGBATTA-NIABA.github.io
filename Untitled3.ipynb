```python

import os
import random
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import tensorflow as tf

from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.utils import to_categorical

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow :", tf.__version__)
print("GPU disponible :", tf.config.list_physical_devices("GPU"))
```

```python

def conv_block(x, filters, dropout_rate):

    # Première convolution
    x = layers.Conv2D(
        filters,
        kernel_size=(3, 3),
        padding="same",
        use_bias=False,
        kernel_regularizer=regularizers.l2(1e-4)
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)

    # Deuxième convolution
    x = layers.Conv2D(
        filters,
        kernel_size=(3, 3),
        padding="same",
        use_bias=False,
        kernel_regularizer=regularizers.l2(1e-4)
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)

    # Réduction des dimensions
    x = layers.MaxPooling2D(pool_size=(2, 2))(x)

    # Réduction du surapprentissage
    x = layers.Dropout(dropout_rate)(x)

    return x

print("Bloc convolutif créé avec succès !")
```

```python
IMG_SIZE = 64

def build_model():

    # Image d'entrée : 64 x 64 en niveaux de gris
    inp = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 1))

    # Augmentation des données
    x = layers.RandomFlip("horizontal")(inp)
    x = layers.RandomRotation(0.05)(x)
    x = layers.RandomZoom(0.10)(x)
    x = layers.RandomTranslation(0.05, 0.05)(x)

    # Blocs convolutifs
    x = conv_block(x, 32, 0.20)
    x = conv_block(x, 64, 0.25)
    x = conv_block(x, 128, 0.30)
    x = conv_block(x, 256, 0.35)

    # Extraction finale des caractéristiques
    x = layers.GlobalAveragePooling2D()(x)

    # Couche entièrement connectée
    x = layers.Dense(
        256,
        activation="relu",
        kernel_regularizer=regularizers.l2(1e-4)
    )(x)

    x = layers.Dropout(0.5)(x)

    # Classification des 7 émotions
    out = layers.Dense(7, activation="softmax")(x)

    model = keras.Model(
        inputs=inp,
        outputs=out,
        name="CNN_RAF_DB"
    )

    return model

model = build_model()

print("Modèle CNN créé avec succès !")
```

```python

model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)

print("Compilation terminée !")
```

```python
import pandas as pd
import numpy as np
import cv2
import os
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.utils import to_categorical

# 1. Chargement des fichiers CSV
# Le dossier RAF_DB contient les fichiers de labels et le dossier DATASET
base_dir = "/content/RAF_DB"
train_df = pd.read_csv(os.path.join(base_dir, "train_labels.csv"))
test_df = pd.read_csv(os.path.join(base_dir, "test_labels.csv"))

# Les colonnes attendues sont généralement le nom de l'image et son étiquette (0 à 6 pour 7 émotions)
print("Aperçu train_labels :\n", train_df.head())

def load_dataset(df, base_path, img_size=64):
    images = []
    labels = []
    # On recherche récursivement ou directement les images dans le dossier DATASET
    for idx, row in df.iterrows():
        # Adapter selon la structure réelle de votre dataset (ex: images dans DATASET/train ou DATASET/test)
        # Ici on cherche l'image dans tout le dossier DATASET
        img_name = row.iloc[0]
        label = row.iloc[1]

        # Adaptation de l'indexation de l'émotion si nécessaire (si vos labels commencent à 1, on fait -1)
        # Généralement dans RAF-DB, les labels vont de 1 à 7. On les ramène de 0 à 6 :
        adjusted_label = int(label) - 1 if int(label) >= 1 else int(label)

        img_path = None
        for root, dirs, files_list in os.walk(os.path.join(base_path, "DATASET")):
            if img_name in files_list:
                img_path = os.path.join(root, img_name)
                break

        if img_path and os.path.exists(img_path):
            # Lecture en niveaux de gris (grayscale)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            # Redimensionnement à IMG_SIZE (64x64)
            img = cv2.resize(img, (img_size, img_size))
            images.append(img)
            labels.append(adjusted_label)

    return np.array(images, dtype="float32"), np.array(labels, dtype="int32")

print("Chargement et prétraitement des images en cours (cela peut prendre un moment)... ")
# Chargement complet
X_train_raw, y_train_raw = load_dataset(train_df, base_dir, IMG_SIZE)
X_val, y_val_raw = load_dataset(test_df, base_dir, IMG_SIZE)

# Normalisation des pixels [0, 255] -> [0, 1]
X_train_raw = X_train_raw / 255.0
X_val = X_val / 255.0

# Ajout de la dimension du canal (grayscale -> shape: (N, 64, 64, 1))
X_train_raw = np.expand_dims(X_train_raw, -1)
X_val = np.expand_dims(X_val, -1)

# Encodage One-Hot pour la classification multi-classe (7 émotions)
Y_train_raw = to_categorical(y_train_raw, num_classes=7)
Y_val = to_categorical(y_val_raw, num_classes=7)

# Division de train_raw pour obtenir X_train et X_val réels si vous n'utilisez pas directement le test_set comme validation
X_train, X_test, Y_train, Y_test = train_test_split(X_train_raw, Y_train_raw, test_size=0.1, random_state=SEED)

# Calcul automatique des poids de classe pour gérer le déséquilibre du dataset
y_train_labels = np.argmax(Y_train, axis=1)
classes = np.unique(y_train_labels)
weights = compute_class_weight(class_weight="balanced", classes=classes, y=y_train_labels)
class_weight = {c: w for c, w in zip(classes, weights)}

print("Chargement terminé !")
print(f"Données d'entraînement : {X_train.shape}")
print(f"Données de validation : {X_val.shape}")
print("Poids des classes :", class_weight)
```

```python
history = model.fit(
    X_train,
    Y_train,
    validation_data=(X_val, Y_val),
    epochs=80,
    batch_size=64,
    class_weight=class_weight,
    callbacks=cbs,
    verbose=1
)
```

```python

face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

print("Détecteur de visage chargé avec succès !")
```

```python

def preparer_visage(image_path):

    img = cv2.imread(image_path)

    if img is None:
        return None, None

    # Conversion en niveaux de gris
    gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Détection des visages
    faces = face_detector.detectMultiScale(
        gris,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    if len(faces) == 0:
        return None, None

    # Sélection du plus grand visage
    x, y, w, h = max(
        faces,
        key=lambda f: f[2] * f[3]
    )

    # Recadrage du visage
    visage = gris[y:y+h, x:x+w]

    # Redimensionnement en 64x64
    visage = cv2.resize(
        visage,
        (IMG_SIZE, IMG_SIZE)
    )

    # Normalisation
    visage = visage.astype("float32") / 255.0

    # Ajout des dimensions nécessaires
    visage = visage[np.newaxis, ..., np.newaxis]

    return visage, (x, y, w, h)

print("Fonction de préparation créée !")
```

```python

def predire_emotion(x):

    # Prédiction sur l'image originale
    prediction1 = model.predict(x, verbose=0)[0]

    # Prédiction sur l'image retournée horizontalement
    image_miroir = x[:, :, ::-1, :]

    prediction2 = model.predict(
        image_miroir,
        verbose=0
    )[0]

    # Moyenne des deux prédictions
    probabilites = (prediction1 + prediction2) / 2

    emotion_index = np.argmax(probabilites)

    emotion = CLASSES[emotion_index]

    confiance = probabilites[emotion_index] * 100

    return emotion, confiance, probabilites

print("Fonction de prédiction créée !")
```

```python
resultats = []

# Définition des classes autorisées en minuscules
CLASSES = [emotion.lower() for emotion in emotion_labels]

# Dictionnaire de correspondance pour aligner le nom du fichier avec la classe
mapping_emotions = {
    "angry": "anger",
    "happy": "happiness",
    "sad": "sadness"
}

for nom in sorted(uploaded.keys()):
    # Nom de base du fichier en minuscules
    nom_base = os.path.splitext(nom)[0].lower()

    # Application de la correspondance
    attendu = mapping_emotions.get(nom_base, nom_base)

    if attendu not in CLASSES:
        print("Nom de fichier non reconnu :", nom)
        continue

    # Préparation du visage
    x, box = preparer_visage(nom)

    if x is None:
        print("Aucun visage détecté dans :", nom)
        continue

    # Prédiction
    predit, confiance, probabilites = predire_emotion(x)

    # Vérification
    correct = (attendu == predit)

    resultats.append({
        "Photo": nom,
        "Emotion attendue": attendu,
        "Emotion prédite": predit,
        "Confiance (%)": round(float(confiance), 2),
        "Résultat": "Correct" if correct else "Incorrect"
    })

    print(
        f"{nom} | Attendue : {attendu:<9} | "
        f"Prédite : {predit:<9} | "
        f"Confiance : {confiance:.2f}%"
    )

print("\nTests terminés !")
```
